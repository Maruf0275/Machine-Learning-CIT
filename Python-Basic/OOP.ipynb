## OOP Advance

### Abstraction

```python
class car:
  def __init__(self):
    self.acc = False
    self.brk = False
    self.clutch = False
  def start(self):
    self.clutch = True
    self.acc = True
    print('Car Started...')

car1 = car()
car1.start()
```

### Encapsulation

```python
class Account:
  def __init__(self, acc_no, acc_pass):
    self.acc_no = acc_no
    self.__acc_pass = acc_pass
  def reset_pass(self):
    print(self.__acc_pass)

acc1 = Account(123456789, '1234')
print(acc1.acc_no)
print(acc1.reset_pass())
```

```python
class Person:
  __name = "Maruf"

  def __hello(self):
    print(f"Hello {self.__name}")
  def welcome(self):
    self.__hello()

p1 = Person()
print(p1.welcome())
```

### Inheritance

```python
class Animal:
    color = 'white'
    print("Animal Sounds...")
class Dog(Animal):
    print("Dog Barking...")
class Cat(Animal):
    print("Cat Meowing...")

d1 = Dog()
print(d1.color)

c1 = Cat()
```

```python
class A:
  varA = 'Welcome to class A'
class B:
  varB = 'Welcome to class B'
class C(A,B):
  varC = 'Welcome to class C'

c1 = C()
print(c1.varA)
print(c1.varB)
print(c1.varC)
```

### Polymorphism

```python
class Car:
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model

  def move(self):
    print("Drive!")

class Boat:
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model

  def move(self):
    print("Sail!")

class Plane:
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model

  def move(self):
    print("Fly!")

car1 = Car("Ford", "Mustang")       #Create a Car object
boat1 = Boat("Ibiza", "Touring 20") #Create a Boat object
plane1 = Plane("Boeing", "747")     #Create a Plane object

for x in (car1, boat1, plane1):
  x.move()
```

### Project Using OOP

```python
from abc import ABC, abstractmethod


# 1. Abstraction
class PaymentSystem(ABC):

    @abstractmethod
    def process_payment(self, amount):
        pass


# 2. Encapsulation
class Account:
    def __init__(self, account_holder, balance=50000):
        self.__account_holder = account_holder
        self.__balance = balance

    # Getter
    def get_balance(self):
        return self.__balance

    # Setter
    def set_balance(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            return True
        return False


# 3. Inheritance
class BkashPayment(PaymentSystem):
    def __init__(self, account):
        self.account = account

    # 4. Polymorphism
    def process_payment(self, amount):
        charge = amount * 0.015
        total = amount + charge

        if self.account.withdraw(total):
            print(f"Bkash Payment: {amount:.2f}")
            print(f"Charge (1.5%): {charge:.2f}")
            print(f"Total Deducted: {total:.2f}")
        else:
            print("Insufficient Balance!")


class CreditCardPayment(PaymentSystem):
    def __init__(self, account):
        self.account = account

    # 4. Polymorphism
    def process_payment(self, amount):
        if amount > 1000:
            discount = amount * 0.10
            total = amount - discount
        else:
            discount = 0
            total = amount

        if self.account.withdraw(total):
            print(f"Credit Card Payment: {amount:.2f}")
            print(f"Discount (10%): {discount:.2f}")
            print(f"Total Deducted: {total:.2f}")
        else:
            print("Insufficient Balance!")


# Main Program

account = Account("Student", 5000)

bkash = BkashPayment(account)
credit_card = CreditCardPayment(account)


# Polymorphism
def make_payment(payment_method, amount):
    payment_method.process_payment(amount)


print("Initial Balance:", account.get_balance())
print()


# 1. Bkash payment
make_payment(bkash, 2000)
print("Remaining Balance:", account.get_balance())
print()


# 2. Credit Card payment
make_payment(credit_card, 1500)
print("Remaining Balance:", account.get_balance())
```
