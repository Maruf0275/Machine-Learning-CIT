```python
class Student:
    university = 'Dit'                  # Class Veriable
    def __init__(self, name, roll):
      self.name = name                  # Instance Veriable
      self.roll = roll
      print(f'{self.name} object created!')

    def display(self):                  # Instance Method
      print(self.name)
      print(self.roll)

    @classmethod
    def change_university(cls, name):   # Class Method
      cls.university = name
```

```python
s3 = Student('Maruf',19)
```

```python
s3.display()
```

```python
print(s3.university)
```

```python
s3.change_university('Dhaka International University')
print(s3.university)
```

```python
class calculator:
  @staticmethod     # Static Method
  def add(a,b):
    return a + b
```

```python
print(calculator.add(10,20))
```

```python
# Create a class
class Person:
  def __init__(self, name, age = 23):
    self.name = name
    self.age = age

  def greet(self):
    print(f"Hello, my name is {self.name}")

# Create an object
p1 = Person('John',39)
p2 = Person('Maruf')
# Call the greet method
p1.greet()
p2.greet()
```

```python
print(p1.name)
```

```python
print(p1.age)
```

```python
print(p2.age)
```
