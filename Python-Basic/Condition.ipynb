```python
mark = int(input('Enter Your Mark: '))

if mark >= 80:
  print('A+')
elif mark >= 70:
  print('A')
elif mark >= 60:
  print('A-')
elif mark >= 50:
  print('B')
elif mark >= 40:
  print('C')
else:
  print('Fail')
```
