```python
def sum_num(*a):   #arguments  (args)
    return a[3]

sum_num(5,4,8,6,54,56,2,6,8,63,6)
```

```python
def num(**a):  # key word arguments (kwrgs)
  return a
num(a = 3, b = 4, c = 1)
```
