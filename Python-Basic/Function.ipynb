```python
def sum_num(*a):   #arguments  (args)
    return a[3]

sum_num(5,4,8,6,54,56,2,6,8,63,6)
```

```python
def num(**a):  # key word arguments (kwrgs)
  return a
num(a = 3, b = 4, c = 1)
```

```python
# Lambda Function

squre = lambda x : x * x

print(squre(5))
```

```python
# Recursive Function
def fact(n):
  if n == 1:
    return 1
  return n * fact(n - 1)

fact(5)
```

```python
def countdown(n):
  if n <= 0:
    print("Done!")
  else:
    print(n)
    countdown(n - 1)

countdown(5)
```

```python
def fibonacci(n):
  if n <= 1:
    return n
  else:
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(7))
```

```python
# Generator

def my_generator():
  yield 1
  yield 2
  yield 3

g = my_generator()
print(next(g))
```

```python
def count_up_to(n):
  count = 1
  while count <= n:
    yield count
    count += 1

for num in count_up_to(5):
  print(num)
```

```python
def simple_gen():
  yield "Emil"
  yield "Tobias"
  yield "Linus"

gen = simple_gen()
print(next(gen))
print(next(gen))
print(next(gen))
```

```python
# Decorator
def changecase(func):
  def myinner():
    return func().upper()
  return myinner

@changecase
def myfunction():
  return "Hello Sally"

print(myfunction())
```
