## Set

```python
a = {1,2,3}
a.add(4)
a
```

```python
a.remove(2)
a
```

```python
s = {1,2,3,4,5,'maruf',70.5,False,True}
s
```

```python
a = {1,2,3,4,5}
b = {4,5,6,7,8}

print(a.union(b))
print(a | b)

print(a.intersection(b))
print(a & b)

print(a.difference(b))
print(a - b)

print(a.symmetric_difference(b))
print(a ^ b)
```

```python
# Empty set declairation
dummy = set()
print(type(dummy))
```

```python
a = {1,2,3,4,5}
print(5 in a)
```
