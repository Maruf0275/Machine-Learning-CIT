## Try Except

```python
try:
  a = int(input("Enter the number: "))
  print(f"Multiplication table of {a} is: ")
  i = 1
  while i <= 10:
    print(f"{a} * {i} = {a * i}")
    i+=1
except Exception as e:
  print(e)
print("Over")
```

```python
# Finally Keyword
try:
  l = [1,2,3,4]
  i = int(input("Enter the index number: "))
  print(l[i])
except:
  print("Some Error Occured")
finally:
  print("I am always Executed")
```

## While loop

```python
i = 0
while(i<=3):
  print(i)
  i+=1
print('Done with the loop')
```

```python
i = int(input('Enter the number: '))
while(i<=30):
  i = int(input('Enter the number: '))
  print(i)
print('Done the loop')
```

```python
count = 5
while(count > 0):
  print(count)
  count-=1
```

```python
car = ['BMW','AUDI','TATA','HONDA']
color = ['RED','BLUE','WHITE','BLACK']
for i in car:
  for j in color:
      print(i,j)
```

```python
a = 5
for i in range(1,11):
  print(f"{a} * {i} = {a*i}")
```

```python
for i in [5,6,7]:
  for j in range(1,11):
    print(f"{i} * {j} = {i*j}")
  print("----------------")
```

```python
# leap year
year = int(input("Enter the year: "))
if(year%4==0):
  if(year%100==0):
    if(year%400==0):
      print(f"{year} is a leap year")
    else:
      print(f"{year} is not a leap year")
  else:
    print(f"{year} is a leap year")
else:
  print(f"{year} is not a leap year")
```

```python
year = int(input("Enter the year: "))
if(year%4==0 and year%100!=0 or year%400==0):
  print(f"{year} is a leap year")
else:
  print(f"{year} is not a leap year")
```

```python
a = [1,2,3,4,5,6,7,8,9]
b =[]
for i in a:
  if(i%2!=0):
    b.append(i)
    a.remove(i)
print(a)
print(b)
```

```python
# list comprehension
l = [i for i in range(1,11) if(i%2!=0)]
print(l)
```

```python
a = ['odd' if(i%2!=0) else 'even' for i in range(1,11)]
print(a)
```
