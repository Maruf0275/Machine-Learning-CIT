##List

```python
data = [1,5,6,3,4,2,8,7,9]

print(data[::-1])
print(sorted(data))
print(data)
```

```python
print(sorted(data)[::-1])
```

```python
data.append(10)
print(data)
```

```python
data.insert(1,11)
print(data)
```

```python
data.pop()
print(data)
```

```python
data.remove(11)
print(data)
```

```python
print(len(data))
```

```python
print(data.count(1))
```

```python
print(data)
print(sum(data))
```

##Tuple

```python
atuple = (10,20,30)
print(atuple)
```

```python
# Tuple packing
name = 'maruf',20,'nibir'
print(name)
```

```python
# Concatenate
a = (10,50,10)
b = (90, )
a += b
print(a)
```

```python
# type casting
a = (10,50,10)
b = list(a)

b.append(90)

a = tuple(b)
print(a)
```
