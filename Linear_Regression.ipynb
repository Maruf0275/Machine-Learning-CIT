```python
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Dataset/placement.csv')
```

```python
df.head()
```

```python
plt.scatter(df['cgpa'], df['package'])
plt.xlabel('CGPA')
plt.ylabel('Package')
```

```python
x = df.iloc[:, 0:1]
y = df.iloc[:, -1]
```

```python
# Split the placement dataset into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# Instantiate and fit the model
lr = LinearRegression()
lr.fit(x_train, y_train)

# Plot the training data scatter and the regression line
plt.figure(figsize=(8, 4))
plt.scatter(df['cgpa'], df['package'], color='blue', alpha=0.6, label='Data points')
plt.plot(x_train, lr.predict(x_train), color='red', linewidth=2, label='Regression Line')
plt.xlabel('CGPA')
plt.ylabel('Package (in Lpa)')
plt.title('CGPA vs Package Linear Regression')
plt.legend()
plt.show()
```

```python
lr.score(x_train, y_train)
```

```python
y_pred = lr.predict(x_test)
y_pred
```

```python
y_test.values
```

```python
# Generate predictions on the test set first
y_pred = lr.predict(x_test)

# Calculate evaluation metrics using the test labels and test predictions
print('MAE:', mean_absolute_error(y_test, y_pred))
print('MSE:', mean_squared_error(y_test, y_pred))
print('R2 Score:', r2_score(y_test, y_pred))
```

```python
# Plot the testing data scatter and the regression line
plt.figure(figsize=(8, 4))
plt.scatter(df['cgpa'], df['package'], color='blue', alpha=0.6, label='Data points')
plt.plot(x_test, y_pred, color='red', linewidth=2, label='Regression Line')
plt.xlabel('CGPA')
plt.ylabel('Package (in Lpa)')
plt.title('CGPA vs Package Linear Regression')
```
