```python
import pandas as pd
```

```python
df = pd.read_csv('/content/drive/MyDrive/Machine Learning(CIT)/Datasets/Breast cancer dataset.csv')
```

```python
df.head()
```

```python
df.drop(columns=['id','Unnamed: 32'],inplace= True)
df.head()
```

```python
df.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(df.drop(columns=['diagnosis']),df['diagnosis'],test_size=0.2,random_state=42)
```

```python
x_train.head()
```

```python
y_train.head()
```

```python
x_train.shape
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
x_train
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=5)
```

```python
knn.fit(x_train,y_train)
```

```python
from sklearn.metrics import accuracy_score
y_pred = knn.predict(x_test)
accuracy_score(y_test,y_pred)
```

```python
import matplotlib.pyplot as plt

accuracies = []

for k in range(1, 16):
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(x_train, y_train)
    y_pred = knn.predict(x_test)
    acc = accuracy_score(y_test, y_pred)
    accuracies.append(acc)
    print(f"k = {k:2d} -> Accuracy: {acc:.4f}")
```

```python
plt.figure(figsize=(8, 4))
plt.plot(range(1, 16), accuracies, marker='o', linestyle='dashed', color='blue', markerfacecolor='red', markersize=8)
plt.title('Accuracy vs. K Value')
plt.xlabel('K')
plt.ylabel('Accuracy')
plt.xticks(range(1, 16))
plt.grid(True)
plt.show()
```
