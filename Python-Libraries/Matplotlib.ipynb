```python
import matplotlib.pyplot as plt
```

```python
# Line Plot
price = [10,17,20,22]
quantity = [5,8,12,17]

plt.figure(figsize=(5,3))
plt.plot(price,quantity,color = 'r',marker = 'o')
```

```python
plt.figure(figsize=(5,3))
plt.plot(price,quantity,color = 'r',marker = 'o')
plt.xlabel('price')
plt.ylabel('quantity')
plt.title('Relation Line Plot')
plt.gca().set_facecolor('lightblue')
plt.grid()
```

```python
plt.figure(figsize=(5,3))
plt.plot(price,quantity,'ro') # 'ro' use for removing line
plt.xlabel('price')
plt.ylabel('quantity')
plt.title('Relation Line Plot')
plt.gca().set_facecolor('lightblue')
plt.grid()
```

```python
plt.figure(figsize=(5,3))
plt.plot(price,quantity,'^k:')
plt.xlabel('price')
plt.ylabel('quantity')
plt.title('Relation Line Plot')
```

```python
plt.figure(figsize=(5,3),facecolor='black')
plt.plot(price,quantity,'^k:')
plt.xlabel('price',color='white')
plt.ylabel('quantity',color='white')
plt.title('Relation Line Plot',color='white')
```

```python
# Bar Plot
price = [10,17,20,22]
quantity = [5,8,12,17]

plt.figure(figsize=(5,3))
plt.bar(price,quantity,color='r')
```

```python
price = [10,20,30,40]
fruit_name = ['Mango','Apple','Banana','Grapes']
colors = ['red','green','yellow','blue']

plt.figure(figsize=(5,3))
plt.bar(fruit_name,price,color=colors)
plt.xlabel('Fruit Name')
plt.ylabel('Price')
plt.title('Fruit Price Range')

plt.ylim(0,60)
```

```python
# Scatter Plot
price = [10,20,30,40]
fruit_name = ['Mango','Apple','Banana','Grapes']

plt.figure(figsize=(5,3))
plt.scatter(fruit_name,price,color='r',s = 100,marker='*',alpha=0.5)

for x,y in zip(fruit_name,price):
    plt.annotate(f'({x},{y})',(x,y))
```

```python
# Sub Plot
plt.figure(figsize=(10,4))
plt.subplot(1,3,1)
plt.plot(price,quantity,color='green',marker = 'o')

plt.subplot(1,3,2)
plt.bar(price,quantity,color=colors)

plt.subplot(1,3,3)
plt.scatter(price,quantity,color='green',marker='*',s=100)

plt.suptitle('Price vs quantity')
plt.savefig('plot.png')
```

```python
# Pie Chart
plt.figure(figsize=(5,3))
plt.pie(price,labels=quantity,autopct='%1.1f%%',startangle=90,shadow=True)
plt.title('Pie Chart')
```

```python
# Multiple Lines
price = [10,17,20,22]
quantity = [5,8,12,17]

plt.figure(figsize=(5,3))
plt.plot(price, color = 'r')
plt.plot(quantity, color = 'green')
plt.title('Multiple Line Plot', loc = 'left')
plt.grid(color = 'green', linestyle = '--', linewidth = 0.5)
```
